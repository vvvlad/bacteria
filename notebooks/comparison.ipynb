# Comparison of Swelling Dynamics: Control vs. Protein Synthesis Arrested

This notebook loads the tracking results from both the control experiment and the protein synthesis arrested experiment to visualize and compare their osmotic swelling dynamics on shared axes.

```python
import pandas as pd
import numpy as np
from plotly.subplots import make_subplots
import plotly.graph_objects as go

import sys
from pathlib import Path

sys.path.insert(0, "../src")

import plotly.io as pio
pio.renderers.default = "notebook_connected"  # nbconvert-friendly plotly output

from IPython.display import display, Markdown

from cell_analysis import (
    show_with_source,
    plot_tension_comparison,
    add_geometry,
)
```

## 1. Load Datasets

Update the paths below to match the exact locations of your generated `tracked.csv` and `track_stats.csv` files for both conditions.

```python
# Load Control Experiment Data
tracked_ctrl = pd.read_csv("../results/control_experiment/extraction/tracked_cells.csv")
stats_ctrl = pd.read_csv("../results/control_experiment/extraction/track_statistics.csv")

# Load Protein Synthesis Arrested Experiment Data
tracked_cam = pd.read_csv("../results/protein_synthesis_arrested/extraction/tracked_cells.csv")
stats_cam = pd.read_csv("../results/protein_synthesis_arrested/extraction/track_statistics.csv")

grad_path_ctrl = "../data/Control-experiment/Gradient Control.xlsx"
grad_df_ctrl = pd.read_excel(grad_path_ctrl).rename(columns={"Frame #": "frame"})
tracked_ctrl = tracked_ctrl.merge(grad_df_ctrl, on="frame", how="left")

grad_path_cam = "../data/Protein-synthesis-arrested/Gradient Control.xlsx"
grad_df_cam = pd.read_excel(grad_path_cam).rename(columns={"Frame #": "frame"})
tracked_cam = tracked_cam.merge(grad_df_cam, on="frame", how="left")
```

```python
# Scaling
PIXEL_SIZE_UM = 0.0645
# BASELINE_FRAMES = 3

tracked_ctrl, stats_ctrl = add_geometry(tracked_ctrl, stats_ctrl, pixel_size_um=PIXEL_SIZE_UM)
tracked_cam, stats_cam = add_geometry(tracked_cam, stats_cam, pixel_size_um=PIXEL_SIZE_UM)
```

## 2. Generate the Tension Plot

Execute the function below using the loaded datasets to display the final comparison graph.

```python
show_with_source(plot_tension_comparison, tracked_ctrl, stats_ctrl, tracked_cam, stats_cam,
                 ctrl_max_frame=25, arr_max_frame=22)
```
